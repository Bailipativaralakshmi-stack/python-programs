```python
# creating file on w mode
file = open("new.txt","w")
file.write("this is my new file")
print("created")
file.close()
```

```python
# file on read mode
file = open("new.txt","r")
data = file.read()
print(data)
file.close()
```

```python
#file on append mode
with open("new.txt","a") as f:
  f.writelines(["\napple","\nbanana","\norange"])
```

```python
with open("new.txt","r") as f:
  print(f.readline())
```

```python
with open("new.txt","r") as f:
  print(f.readlines())
```

```python
#write a program that copy one file to another
with open("new.txt","r") as f:
  with open("new1.txt","w") as f1:
    for line in f:
      f1.write(line)
    print("copied")
```

```python
with open("new1.txt","r") as f1:
  data = f1.read()
  print(data)
```

```python
#that each line of a file in reversee order
file = open("new.txt","r")
data = file.readlines()
for i in data:
  print(i[::-1])
```

```python
# copy the words into another file and sorted and lowercase
with open("new.txt","r") as f:
  data = f.read().split()
  data_lower = [i.lower() for i in data]
  sort_data=sorted(data_lower)
with open("f4.txt","w") as f2:
  for i in sort_data:
    f2.write(i+"\n")
  print("done")
```

```python
#write a python program print lines and characters
with open("new.txt","r") as f:
  data = f.readlines()
for i,line in enumerate(data,1):
  print(i,len(line))

```

```python
#binary files
import pickle
with open("f1.dat","wb") as f:
  data = [1,2,3,"hello"]
  pickle.dump(data,f)
  print("done")
```
