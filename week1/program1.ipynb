```python
#arthmetic operators
a = int(input("enter a number:"))
b = int(input("enter another number:"))
print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a%b)
print(a//b)
print(a**b)
```

```python
#comparison operators
a = int(input("enter a number:"))
b = int(input("enter another number:"))
print(a==b)
print(a>=b)
print(a<=b)
print(a>b)
print(a<b)
print(a!=b)

```

```python
#logical operators
a = int(input("enter a number:"))
b = int(input("enter another number:"))
print(a>b and b>a)
print(a==b or b>=a)
print(not(a>b))
```
