```python
#built-in operstions
d = {"sno":8,"name":"vara","marks":98,"gpa":9.7}
print(len(d))
print(sorted(d))
print(max(d))
print(min(d))
```

```python
# dict methods
d = {"sno":8,"name":"vara","marks":98,"gpa":9.7}
print(d.get("name"))
print(d.keys())
print(d.values())
print(d.items())
```

```python
#dict comprenssion
d = {i:i*i for i in range(1,6)}
print(d)
```

```python
#dict runtime input
k = [i for i in input().split()]
v = [int(i) for i in input().split()]
d = {i:j for i,j in zip(k,v)}
print(d)
```
