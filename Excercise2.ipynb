```python
# maximum among three numbers
a = int(input("enter first number"))
b = int(input("enter second number"))
c = int(input("enter third number"))
if a>b and a>c:
  print(f"{a} is maximum")
elif b>a and b>c:
  print(f"{b} is maximum")
else:
  print(f"{c} is maximum")
```

```python
#check given year is leap or not
year = int(input("enter a year"))
if year%400 == 0 or year%4 == 0 and year%100 != 0:
  print(f"{year} is leap")
else:
  print(f"{year} is not leap")
```

```python
# example code on match code
day = int(input("enter a day"))
match day:
  case 1:
    print("monday")
  case 2:
    print("tuesday")
  case 3:
    print("wednesday")
  case 4:
    print("thursday")
  case 5:
    print("friday")
  case 6:
    print("saturday")
  case 7:
    print("sunday")
  case _:
    print("invalid day")

```

```python
#program that finds sum of the digits of a given number
num = int(input("enter a number"))
sum = 0
while(num!=0):
  r = num%10
  sum = sum + r
  num = num//10
print(sum)
```

```python
# reverse of a number
num = int(input("enter a number"))
sum = 0
while(num!=0):
  r = num%10
  sum = sum *10 + r
  num = num//10
print(sum)
```

```python
n = int(input("enter a number"))
if n>1:
  for i in range(2,n):
    if n%i == 0:
      print("not prime")
      break
  else:
    print("prime")



```

```python
# prime numbers between the range
a = int(input("enter starting num:"))
b = int(input("enter ending num:"))
for i in range(a,b+1):
  for j in range(2,i):
    if(i%j == 0):
      break
  else:
    print(i)
```
