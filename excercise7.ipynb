```python
#set
s1 = {1,5,8,9,4}
s2 = {2,4,6,10}
print(s1|s2)
print(s1 & s2)
print(s1-s2)
print(s1^s2)
```

```python
# set methods
s1 = {1,4,5,6,7}
s2 = {33,55,66,77,88}
s1.add(33)
print(s1)
s1.update((34,22,11))
print(s1)
s2.remove(55)
print(s2)
s2.discard(33)
print(s2)
s1.pop()
print(s1)
```

```python
# set methods
s1 = {2,4,6,7,8}
s2 = {42,4,5,6,7}
print(s1.union(s2))
print(s1.intersection(s2))
print(s1.difference(s2))
print(s1.symmetric_difference(s2))
```
