```python
# cover list operators
a = [1,2,3,"vara"]
b = [3,7,9,"maha"]
print(a+b)
print(a*2)
print(a[3])
print(a[-2])
print(b[0:3])
print(50 not in a)
for x in a:
  print(x)
```

```python
# built in function in list
a = [20,30,40,60,10]
print(len(a))
print(max(a))
print(min(a))
print(sum(a))
print(list(reversed(a)))
print(sorted(a))
```

```python
#built in methods
a = [3,4,5,67,77,88,99,100]
b = [1,11,13,14]
a.append("vara")
print(a)
a.extend(b)
a.insert(0,"ee")
a.pop()
print(a)
print(a.count(4))
print(a.reverse())
print(a)
print(b.sort())
print(b)
print(a.remove(4))
print(a.pop())
print(a.index(88))
print(a)
```

```python
# program that print without using sum()
a = [2,5,4,8,9,10]
sum = 0
for i in a:
  sum = sum+i
print(sum)
```

```python
#print the list elements at odd index
for i in range(len(a)):
  if i%2 != 0:
    print(a[i])
```

```python
# print even elements
for i in range(len(a)):
  if a[i]%2==0:
    print(a[i])
```
